```python
import numpy as np
import matplotlib.pyplot as plt
t=np.loadtxt("./dane_zad_5.txt")
x=[t[i][0] for i in range(len(t))]
y=[t[i][1] for i in range(len(t))]
plt.scatter(x,y)
beta_1,beta_0=np.polyfit(x,y,1)
y_reg=[beta_0 + beta_1 * i for i in x]
plt.plot(x,y_reg)
plt.show()
```

```python
y_temp=np.log(y)
plt.scatter(x,y_temp)
plt.show()
```
