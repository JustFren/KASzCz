```python
import numpy as np
import matplotlib.pyplot as plt
n=100
space=np.linspace(0,10,100)
def gen_Y(n,space):
    res=np.random.normal(0,1,n)
    return ([2+0.5*space[i]+res[i] for i in range(n)],res)
def beta_1(x_vect,y_vect):
    num=0
    den=0
    mean=np.mean(x_vect)
    for i in range(len(x_vect)):
        num+=(x_vect[i] - mean)*y_vect[i]
        den+=(x_vect[i]-mean)**2
    return num/den
def beta_0(x_vect,y_vect):
    return np.mean(y_vect) - beta_1(x_vect,y_vect)*np.mean(x_vect)

def calc_leverage(n,i,space,rand):
    return (1/n)+((rand[i]-np.mean(rand))**2)/(np.sum((rand-np.mean(rand))**2))
def cook_distance(n,space,rand,res):
    Y=[beta_0(space,rand)+beta_1(space,rand)*i for i in space]
    Y_star=[beta_0(space[:-1],rand[:-1])+beta_1(space[:-1],rand[:-1])*i for i in space]
    s_sq=np.sum(res**2)/(n-2)
    return (np.sum((np.add(Y,np.negative(Y_star)))**2)/2*s_sq)
org=gen_Y(n,space)
plt.scatter(space,org[0])
Y=[beta_0(space,org[0])+beta_1(space,org[0])*i for i in space]
plt.plot(space,Y,color="orange")
plt.show()
    
```

```python
temp_Y=np.append(org[0],35)
temp_space=np.append(space,5)
plt.scatter(temp_space,temp_Y)
plt.show()
print("residue",35-(2+0.5*5))
print("levarage",calc_leverage(n,100,temp_space,temp_Y))
print("Cook distance",cook_distance(n,temp_space,temp_Y,org[1]))
```
